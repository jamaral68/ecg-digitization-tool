```python
from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.image import imread
import sys

sys.path.append(str(Path.cwd().parent / "src"))

from ecg_scanner.scanner import ECGScanner

p = Path.cwd().parent / "datasets" / "images"
```

```python
scanner = ECGScanner(
    output_dir=Path.cwd().parent / "datasets" / "new_output_ecg_scanner"
)
for image_path in p.glob("*.jpeg"):
    scanner.scan(image_path)
```

```python
scanner = ECGScanner(
    output_dir=Path.cwd().parent / "datasets" / "new_output_ecg_scanner"
)
for image_path in p.glob("*.jpeg"):
    scanner.scan(image_path)
```

```python
dir1 = p.parent / "new_output_ecg_scanner"
dir2 = p.parent / "old_output_ecg_scanner"
dir3 = p

images1 = sorted(dir1.glob("*.jpeg"))
images2 = sorted(dir2.glob("*.jpeg"))
images3 = sorted(dir3.glob("*.jpeg"))

n = min(len(images1), len(images2), len(images3))
for i in range(n):
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))
    ax1.imshow(imread(images1[i]), cmap="gray")
    ax1.set_title(images1[i].name + " - Versão nova")
    ax1.axis("off")
    ax2.imshow(imread(images2[i]), cmap="gray")
    ax2.set_title(images2[i].name + " - Versão antiga")
    ax2.axis("off")
    ax3.imshow(imread(images3[i]))
    ax3.set_title(images3[i].name + " - Original")
    ax3.axis("off")
    plt.tight_layout()
    plt.show()
```
