```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Carregar imagem
img = cv2.imread("bucket/img20250225_11454133.png")

# Converter para escala de cinza
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# # Criar máscara manual para regiões pretas (valores escuros)
# mask_black = cv2.inRange(gray, 0, 50)

# # Criar máscara manual para letras (escuras mas menores)
# mask_text = cv2.inRange(gray, 0, 120)

# # Unir as máscaras (retângulos + letras)
# mask = cv2.bitwise_or(mask_black, mask_text)

# Teste Manual
mask = np.zeros(gray.shape, dtype=np.uint8)

# Exemplo: retângulo preto no canto inferior esquerdo
mask[gray.shape[0]-400:gray.shape[0]-20, 0:200] = 255  

# Aplicar inpainting
result = cv2.inpaint(img, mask, 3, cv2.INPAINT_TELEA)


# Depois adiciona mais regiões com base nas coordenadas


# Mostrar resultado
plt.figure(figsize=(15,5))
plt.subplot(1,3,1), plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)), plt.title("Original")
plt.subplot(1,3,2), plt.imshow(mask, cmap="gray"), plt.title("Máscara")
plt.subplot(1,3,3), plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB)), plt.title("Resultado")
plt.show()


```
