# 🏷️ Projeto de Anotação Manual ECG com LabelMe

Este projeto permite criar **máscaras binárias** de linhas de ECG com a ferramenta [LabelMe](https://github.com/wkentaro/labelme).

---

## 📁 Estrutura do Projeto

```
labelme_ecg_project/
├── images/          # ECGs originais a serem anotados
├── masks/           # Máscaras PNG geradas (output)
└── README.md        # Este arquivo
```

---

## ✅ Requisitos

- Python 3.7+
- Instalar o LabelMe:

```bash
pip install labelme
```

---

## ✏️ Como Anotar

1. Coloque as imagens dos ECGs em `images/`

2. Rode o LabelMe:
```bash
labelme images/
```

3. Para cada imagem:
   - Clique em **Create Polygons**
   - Desenhe **somente sobre a linha do ECG** (ignorar fundo e grades)
   - Use a label: `linha_ecg`
   - Salve o `.json` (`Ctrl+S`)

---

## 🧪 Como gerar máscaras PNG

Depois de salvar os `.json`, converta as anotações para máscaras com:

```bash
labelme_json_to_dataset images/SEU_ARQUIVO.json -o masks/SEU_ARQUIVO
```

Ou para todas de uma vez:

```bash
for f in images/*.json; do
    labelme_json_to_dataset "$f" -o "masks/$(basename "$f" .json)";
done
```

As máscaras estarão em:
```
masks/NOME_DA_IMAGEM/label.png
```

Você pode renomear e mover os `label.png` para o seu diretório de treino U-Net.
